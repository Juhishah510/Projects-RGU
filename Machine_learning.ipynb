### Dataset Explanation: Default of Credit Card Clients

This dataset contains information on default payments, demographic factors, credit data, history of payment, and bill statements of credit card clients in Taiwan from **April 2005 to September 2005**.

There are **25 variables (columns)** in this dataset:

#### 1. Identification and Demographics
* **`ID`**: ID of each client
* **`LIMIT_BAL`**: Amount of given credit in NT dollars (includes individual and family/supplementary credit)
* **`SEX`**: Gender
  * `1` = male
  * `2` = female
* **`EDUCATION`**: Level of education
  * `1` = graduate school
  * `2` = university
  * `3` = high school
  * `4` = others
  * `5` = unknown
  * `6` = unknown
* **`MARRIAGE`**: Marital status
  * `1` = married
  * `2` = single
  * `3` = others
* **`AGE`**: Age of the client in years

#### 2. History of Past Payment (April - September 2005)
These columns track the monthly repayment status. The scale is as follows: `-1` = pay duly, `1` = payment delay for one month, `2` = payment delay for two months, ... `8` = payment delay for eight months, `9` = payment delay for nine months and above (Note: `0` and `-2` typically represent no consumption or transaction history).
* **`PAY_0`**: Repayment status in September, 2005
* **`PAY_2`**: Repayment status in August, 2005
* **`PAY_3`**: Repayment status in July, 2005
* **`PAY_4`**: Repayment status in June, 2005
* **`PAY_5`**: Repayment status in May, 2005
* **`PAY_6`**: Repayment status in April, 2005

#### 3. Amount of Bill Statement (April - September 2005 in NT dollars)
* **`BILL_AMT1`**: Amount of bill statement in September, 2005
* **`BILL_AMT2`**: Amount of bill statement in August, 2005
* **`BILL_AMT3`**: Amount of bill statement in July, 2005
* **`BILL_AMT4`**: Amount of bill statement in June, 2005
* **`BILL_AMT5`**: Amount of bill statement in May, 2005
* **`BILL_AMT6`**: Amount of bill statement in April, 2005

#### 4. Amount of Previous Payment (April - September 2005 in NT dollars)
* **`PAY_AMT1`**: Amount paid in September, 2005
* **`PAY_AMT2`**: Amount paid in August, 2005
* **`PAY_AMT3`**: Amount paid in July, 2005
* **`PAY_AMT4`**: Amount paid in June, 2005
* **`PAY_AMT5`**: Amount paid in May, 2005
* **`PAY_AMT6`**: Amount paid in April, 2005

#### 5. Target Variable
* **`default payment next month`**: Ground truth / Class label indicating whether the client defaulted
  * `1` = yes (defaulted)
  * `0` = no (did not default)

```python

import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

df["AGE"].hist(ax=axes[0, 0], bins=20)
axes[0, 0].set_title("Customer Age")
axes[0, 0].set_xlabel("Age")

df["LIMIT_BAL"].hist(ax=axes[0, 1], bins=30)
axes[0, 1].set_title("Credit Limit")
axes[0, 1].set_xlabel("Credit limit (NT dollars)")

df["PAY_AMT1"].hist(ax=axes[1, 0], bins=30)
axes[1, 0].set_title("September Payment Amount")
axes[1, 0].set_xlabel("Payment amount (NT dollars)")

df["BILL_AMT1"].hist(ax=axes[1, 1], bins=30)
axes[1, 1].set_title("September Bill Amount")
axes[1, 1].set_xlabel("Bill amount (NT dollars)")

plt.tight_layout()
plt.show()
```
